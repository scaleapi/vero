```python
from tqdm import tqdm
from vero_benchmarking.datasets import (
    AflowDrop,
    AflowGsm8k,
    AflowHotpotqa,
    AflowHumaneval,
    AflowMath,
    AflowMbpp,
)

builders = [AflowDrop, AflowGsm8k, AflowHotpotqa, AflowHumaneval, AflowMath, AflowMbpp]
builders = list(map(lambda x: x(), builders))


datasets = {}

for builder in tqdm(builders):
    datasets[builder.name] = builder.build()
```

```python
for name, ds in datasets.items():
    print(name)
    print(ds)
    print("-" * 100)
```
